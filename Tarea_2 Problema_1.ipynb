```python
import numpy as np

# Importa la biblioteca de numpy
```

```python
rng = np.random.default_rng(seed=800)

# Elige numeros "aleatorios" entre 0 y 800
```

```python
L = [1.0, 2.5, 0.8, 5.0, 10.0]
d = [10, 25, 15, 40, 60]
```

```python
arreglo_L = np.array(L)
arreglo_d = np.array(d)

# Convierte las listas de python en arreglos
```

```python
def calc_flujo(L, d):
    """
    Calcula el flujo de las estrellas.

    Parameters
    ----------
    L = numpy.ndarray
        Datos de luminosidad de cada estrella.
    d = numpy.ndarray
        Datos de distancia de caada estrella

    Returns
    -------
    numpy.ndarray
        Resultado de la formula de flujo para cada dato.
    """
    flujo = L / (4 * np.pi * np.power(d, 2))
    return flujo
    
flujo = calc_flujo(arreglo_L, arreglo_d)
```

```python
print(np.where(flujo == np.max(flujo))) # Imprime el indice del flujo mayor.
```

```python
print(np.where(flujo == np.min(flujo))) # Imprime el indice del flujo menor.
```

```python
print(np.mean(flujo)) # Imprime el promedio de los flujos.
```

```python
print(np.median(flujo)) # Imprime la mediana de flujos.
```

```python
mask = flujo >= 10**-11 # Mascara booleana que muestra solo los flujos mayores a 10^-11
```

```python
flujo[mask] # Imprime el arreglo aplicando la mascara
```

```python
np.flip(flujo[np.argsort(flujo, axis=0)]) # Imprime los datos de flujo ordenados de mayor a menor
```
