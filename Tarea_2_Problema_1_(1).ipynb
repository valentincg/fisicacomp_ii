```python
import numpy as np

# Importa la biblioteca de numpy
```

```python
rng = np.random.default_rng(seed=800)

# Elige numeros "aleatorios" entre 0 y 800
```

```python
L = [1.0, 2.5, 0.8, 5.0, 10.0]
d = [10, 25, 15, 40, 60]
```

```python
arreglo_L = np.array(L)
arreglo_d = np.array(d)

# Convierte las listas de python en arreglos
```

```python
def calc_flujo(L, d):
    """
    Calcula el flujo de las estrellas.

    Parameters
    ----------
    L = numpy.ndarray
        Datos de luminosidad de cada estrella.
    d = numpy.ndarray
        Datos de distancia de caada estrella

    Returns
    -------
    numpy.ndarray
        Resultado de la formula de flujo para cada dato.
    """
    flujo = L / (4 * np.pi * np.power(d, 2))
    return flujo

flujo = calc_flujo(arreglo_L, arreglo_d)
```

```python
print(np.where(flujo == np.max(flujo))) # Imprime el indice del flujo mayor.
```

```python
print(np.where(flujo == np.min(flujo))) # Imprime el indice del flujo menor.
```

```python
print(np.mean(flujo)) # Imprime el promedio de los flujos.
```

```python
print(np.median(flujo)) # Imprime la mediana de flujos.
```

```python
mask = flujo >= 10**-11 # Mascara booleana que muestra solo los flujos mayores a 10^-11
```

```python
flujo[mask] # Imprime el arreglo aplicando la mascara
```

```python
np.flip(flujo[np.argsort(flujo, axis=0)]) # Imprime los datos de flujo ordenados de mayor a menor
```

### PROBLEMA 2

```python
años = np.linspace(0,10, 500) # Crea 500 puntos equidistantes en 10 años.
```

```python
flujo2 = np.ones_like(años) # Crea un arreglo lleno de 1 en todos los tiempos.
```

```python
flujo2 = flujo2 + rng.normal(loc=0.0, scale=0.002, size=años.size) # Agrega el ruido de gauss a todos los tiempos.
```

```python
mascara_transito = (años >= 4.5) & (años <= 5.0) # Mascara que selecciona el intervalo entre 4.5 y 5 años.
```

```python
flujo2[mascara_transito] -= 0.02 # Agrega la disminucion de 0.2 al flujo.
```

```python
promedio_f = np.mean(flujo2)
print(promedio_f) # Imprime el promedio del flujo.
```

```python
desv_standar = np.std(flujo2)
print(desv_standar) # Imprime la desviacion estandar del flujo.
```

```python
print("Flujo minimo:",flujo[(np.argmin(flujo2))],"Indice:", np.argmin(flujo2)) # Imprime el flujo minimo y su indice.
```

```python
print(años[np.argmin(flujo2)]) # Imrprime el tiempo del flujo minimo.
```

```python
condicion = promedio_f - 3 * desv_standar # Condicion para que los puntos tienen que cumplir.
mascara_3sigma = flujo < condicion # Mascara que aplica la condicion.

N_sel = np.sum(mascara_3sigma) # Variable del numero de puntos.
print("Puntos identificados:", N_sel) # Imprime la variable.
```

```python
flujo_cond = flujo[mascara_3sigma] # Variable que muestra el flujo promedio de los transitos.
profundidad = 1 - np.mean(flujo_cond) # Profundidad de los transitos
print("Profundidad de los tranistos:" , profundidad) # Imprime la profundidad promedio.
```

```python
razon_rad = np.sqrt(profundidad) # Variable de la razon de los radios de los planetas y de las estrellas.
print(razon_rad) # Imprime la razon.
```
